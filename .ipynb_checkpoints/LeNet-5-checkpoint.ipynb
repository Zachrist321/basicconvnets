### LeNet-5 Implementation using TensorFlow

```python
import tensorflow as tf
import numpy as np
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# Padding is done to meet the original architecture expectation of 32*32 Zero Padding
x_train = np.pad(x_train, ((0,0), (2,2), (2,2)))
x_test = np.pad(x_test, ((0,0), (2,2), (2, 2)))
```

```python
#Since tf expects 4D input, we have to expand the dims
x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)
```

```python
#Now normalize pixels to be 0-1
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0
```

```python
#Building the LeNet-5 Model using functional API
def lenet_functional_model():
    inputs = layers.Input(shape=(32, 32, 1))
    x = layers.Conv2D(6, kernel_size=(5,5), activation='tanh')(inputs)
    x = layers.AveragePooling2D(pool_size=(2,2))(x)
    x = layers.Conv2D(16, kernel_size=(5,5), activation='tanh')(x)
    x = layers.AveragePooling2D(pool_size=(2,2))(x)
    x = layers.Flatten()(x)
    x = layers.Dense(units=120, activation='tanh')(x)
    x = layers.Dense(units=84, activation='tanh')(x)
    outputs = layers.Dense(units=10, activation='softmax')(x)
    model = models.Model(inputs=inputs, outputs=outputs)
    return model
```

```python
model = lenet_functional_model()
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy']
             )
```

```python
model.summary()
```

```python
history = model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    shuffle=True
)
```

### Lets Evaluate the Model on Test Set

```python
test_loss, test_accuracy = model.evaluate(x_test, y_test)
```

```python
pred = model.predict(x_test[:10])
pred_labels = tf.argmax(pred, axis=1)
```

#### This is the First Novel CNN Architecture paper named 'gradient based learning applied to document recognition'. The prior architecture was Fully Connected Dense Neural Network which treats each pixels single channel as an independent feature. LeNet-5 came up enhancing it by averaging the nearby pixels using convolution, this minimizes parameters to be learned and delivered great accuracy when compared to classical Deep Network. This Algorithm didn't use padding, that is a drawback beyond some computational improvements later made by ResNet, GoogLNet, and MobileNet architectures.
